### 1 ) 🔐 Scenario-Based Exercise: Security Door Access
Scenario:
You are a security engineer at a high-tech facility. To enter the building, employees must type in a secret code on a keypad. The code for today is "123". You need to write a program that checks if the person has entered the correct code.

🧠 Mission Brief:
Your task is to write a program that:

Asks the person to enter the access code.

If the code is "123", the system should print:
"🔓 Access granted. Welcome inside!"

If the code is incorrect, the system should print:
"⛔ Access denied. Incorrect code."



```python
correct_code = "123"
code = (input("Enter the access code: "))         # user input

if correct_code == code:                          # condition:evaluate user code matches the allocated code
  print("🔓 Access granted. Welcome inside!")     # success
else:
  print("⛔ Access denied. Incorrect code.")      # failure
```

**A more complex solution that gives users the clasical three attempts before permanently locking the door:
**

*  FOR LOOP - there is a fixed number of attempts (3), similar to real world
*  Is WHILE LOOP an option? - only if I do not want to limit the users tries and can be a security risk





```python
correct_code = "123"
max_attempts = 3

for attempt in range(max_attempts):                    # runs the loop up to 3 times (attempts 0, 1, and 2)
  code = input("Enter the access code: ")

  if code == correct_code:
    print("🔓 Access granted. Welcome inside!")
    break                                               # code correct then exits the loop, user is not asked again the code.
  else:
    remaining_attempts = max_attempts - (attempt + 1)   # tracking the atempts
    if remaining_attempts > 0:
      print(f"⛔ Access denied. Incorrect code. {remaining_attempts} attempt(s) remaining.")
    else:
      print("🔒 Too many incorrect attempts. Door locked!")
```

### 2) You’re building a simple number-checking tool for a maths learning app. The app should tell users whether their chosen number is divisible by 3 and 7.

🧪 Task:
Write a Python program that:

Asks the user to enter a number.

Checks whether the number is divisible by both 3 and 7.

Displays a clear message with the result.

```python
number = int(input("Enter a number: "))

if number % 3 == 0 and number % 7 == 0:                         # check if a number divides evenly without a remainder.
  print(f"{number} is divisible by both 3 and 7")
else:
  print(f"{number} is not divisible by both 3 and 7")
```

### 3) You're developing a simple calculator tool for a local community learning website. Users should be able to choose what operation they want to perform and continue using the calculator until they decide to exit.

🧪 Task:
Write a Python program that:

Displays a menu with the following options:

* 1- Addition

* 2- Subtraction

* 3- Multiplication

* 4-Division

* 5- Exit

Asks the user to choose an operation

Asks for two numbers, then performs the chosen operation

Repeats the menu until the user selects Exit

```python
option = ("Choose an operation: \n 1- Addition \n 2- Subtraction \n 3- Multiplication \n 4- Division \n 5- Exit")
operation = 1                                               # can be any number 0,1 ... just not 5, that way condition is true and loop will start
while operation != 5:
  print(option)
  operation = int(input("Enter an operation: "))

  if operation == 5:
    print("Exit")
  elif operation == 1:
    num1 = int(input("Enter a number: "))
    num2 = int(input("Enter a number: "))
    print("Result:", num1 + num2)
  elif operation == 2:
    num1 = int(input("Enter a number: "))
    num2 = int(input("Enter a number: "))
    print("Result:", num1 - num2)
  elif operation == 3:
    num1 = int(input("Enter a number: "))
    num2 = int(input("Enter a number: "))
    print("Result:", num1 * num2)
  elif operation == 4:
    num1 = int(input("Enter a number: "))
    num2 = int(input("Enter a number: "))
    print("Result:", num1 / num2)
  else:
    print("Invalid choice. Please pick 1 to 5.")
```

### 4) You’re building a fitness tracker app. The app should calculate the total number of steps someone would take if they increase their steps by 1 each day — starting from 1 up to a number they choose.

🧪 Task:
Write a Python program that:

Asks the user to enter a positive number

Calculates the sum of all numbers from 1 to that number

Displays the total
<p> If the user enters 10, the output should be:

"The sum of numbers from 1 to 10 is 55"

```python
step = int(input("Enter a number: "))
sum_steps = 0
for day in range(1, step + 1):
  sum_steps += day
print(sum_steps)
```

🧪 Task:
Write a Python program that:

Asks the user to enter a positive integer

Checks whether the number is a prime number

Displays a message telling the user whether the number is prime or not



```python
number = int(input("Enter a number: "))
if number > 1:                                  # prime number is a whole number greater than 1 that can only be divided evenly by 1 and itself.
  for i in range(2, number):                    # 2 is first prime number
    if number % i == 0:                         # no remainder condition
      print(f"{number} is not a prime number")
      break
  else:
    print(f"{number} is a prime number")
```

### 6) You’re building a tool for a puzzle game where certain levels require calculating the factorial of a number to unlock a clue. A factorial of a number is the product of all positive integers from 1 up to that number.

For example:
5! = 5 × 4 × 3 × 2 × 1 = 120

🧪 Task:
Write a Python program that:

Asks the user to enter a non-negative integer

Calculates the factorial of that number

Displays the result



```python
number = int(input("Enter a non-negative integer:"))
if number < 0:
  print("Please enter a non-negative integer")
else:
  factorial = 1
  for i in range(1, number + 1):
    factorial *= i
  print(f"The factorial of {number} is {factorial}")
```
